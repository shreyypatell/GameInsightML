```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../dataset/vgsales.csv')
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df['Genre'].value_counts().plot(kind='bar', figsize=(10,5))
plt.title('Games per Genre')
plt.show()
```

```python
df.groupby('Year')['Global_Sales'].sum().plot(figsize=(10,5))
plt.title('Global Sales by Year')
plt.show()
```

```python
corr = df.select_dtypes('number').corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```
